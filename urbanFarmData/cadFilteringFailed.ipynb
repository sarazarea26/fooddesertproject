# CAD Filtering 

```python
import requests
import geopandas as gpd
import pandas as pd
import os
from shapely.geometry import Point
```

## Dallas

```python
base_url = 'https://services2.arcgis.com/rwnOSbfKSwyTBcwN/arcgis/rest/services/DallasTaxParcels/FeatureServer/0/query'
OUTPUT_FILE = 'dallas_parcels.gpkg'

if os.path.exists(OUTPUT_FILE):
    os.remove(OUTPUT_FILE)

offset = 0
record_limit = 1000 
is_fetching = True

while is_fetching:
    params = {
        'where': 'AREA_FEET <= 217800',
        'outFields': 'OBJECTID, AREA_FEET',
        'outSR': '4326',
        'f': 'geojson',
        'resultOffset': offset,
        'resultRecordCount': record_limit,
        'returnGeometry': 'true'
    }

    response = requests.get(base_url, params=params)

    if response.status_code == 200:
        data = response.json()
        features = data.get('features', [])

        if len(features) > 0:
            geojson_batch = {'type': 'FeatureCollection', 'features': features}
            batch_gdf = gpd.GeoDataFrame.from_features(geojson_batch, crs='EPSG:4326')

            batch_gdf.to_file(OUTPUT_FILE, layer='parcels', driver='GPKG', mode='a')

            print(f'Streamed records {offset} to {offset + len(features)} directly to storage.')
            offset += record_limit
            
        else: 
            is_fetching = False
            print(f'Extraction complete. File saved to {OUTPUT_FILE}')

    else: 
        print(f'HTTP Error: {respons.status_code}')
        is_fetching = False
```

```python
# Filtering 
df = pd.read_csv('urbanFarmData/pre_llm_datasets/filtered_dallas_farms.csv')
df.dropna(axis=1, how='all', inplace=True)
geometry = [Point(xy) for xy in zip(df['lon'], df['lat'])]
farm_points = gpd.GeoDataFrame(df, geometry=geometry, crs='EPSG:4326')

cad_parcels = gpd.read_file('urbanFarmData/CAD_filtering/dallas_parcels.gpkg')

cad_parcels = cad_parcels.to_crs('EPSG:4326')

joined_gdf = gpd.sjoin(farm_points, cad_parcels, how='inner', predicate='intersects')

joined_gdf = joined_gdf.set_geometry(cad_parcels.geometry.loc[joined_gdf['index_right']].values)

farms_meters = joined_gdf.to_crs('EPSG:3081')

farms_meters['calculated_acres'] = farms_meters.geometry.area * 0.000247105

under_five_acre_mask = farms_meters['calculated_acres'] <= 5.0

urban_farms_filtered = farms_meters[under_five_acre_mask]

dropped_farms = farms_meters[~under_five_acre_mask]

print(f'{len(dropped_farms)} filtered out')
if not dropped_farms.empty:
    # Adjust 'farm_name' to match your actual column name
    for idx, row in dropped_farms.iterrows():
        print(f"Dropped: {row['name']} | Size: {row['calculated_acres']:.2f} acres")
else:
    print('No farms dropped')

print(f'{len(urban_farms_filtered)} rows under 5 acres.') 
```
