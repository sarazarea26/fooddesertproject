## Diagnostics 

```python
# bexar county 
bexar_data = gpd.read_file("modeling_data/bexar_modeling_final.gpkg")

# dropping NaNs 
variables_to_use = ['distance_to_nearest_uf', 
                 'distance_to_nearest_grocery',
                 'distance_to_nearest_fm',
                 'median_income',
                 'pct_no_vehicle',
                 'median_age',
                 'walking_ind',
                 'E_CHD'
                ]

bexar_clean = bexar_data.dropna(subset=variables_to_use).copy()

# logging income 
bexar_clean['log_income'] = np.log(bexar_clean['median_income'])

# generating spatial weights 
w = weights.distance.KNN.from_dataframe(bexar_clean, k=8)
# row-standardization
w.transform = "R"
```

```python
# Running OLS Model 
ind_variables = ['distance_to_nearest_uf', 
                 'distance_to_nearest_grocery',
                 'distance_to_nearest_fm',
                 'log_income',
                 'pct_no_vehicle',
                 'median_age',
                 'walking_ind'
                ]

ols = spreg.OLS(
    # Dependent variable
    bexar_clean[["E_DIABETES"]].values,
    # Independent variables
    bexar_clean[ind_variables].values,
    # Dependent variable name
    name_y="pct_diabetes",
    # Independent variable name
    name_x=ind_variables,
)
print(ols.summary)

# generating moran's residuals 
moran_residuals = spreg.MoranRes(ols, w, z=True)
print(f"Observed Moran's I:  {moran_residuals.I:.4f}")
print(f"Expected Moran's I:  {moran_residuals.eI:.4f}")
print(f"Standardized Z-score: {moran_residuals.zI:.4f}")
print(f"Analytical P-value:  {moran_residuals.p_norm:.4f}")

# generating LM tests 
lm_results = spreg.LMtests(ols, w)
print("LM Error p-value:", lm_results.lme[1])
print("LM Lag p-value:", lm_results.lml[1])
```

```python
# travis 
travis_data = gpd.read_file("modeling_data/travis_modeling_final.gpkg")

# dropping NaNs 
variables_to_use = ['distance_to_nearest_uf', 
                 'distance_to_nearest_grocery',
                 'distance_to_nearest_fm',
                 'median_income',
                 'pct_no_vehicle',
                 'median_age',
                 'walking_ind',
                 'E_CHD'
                ]

travis_clean = travis_data.dropna(subset=variables_to_use).copy()

# logging income 
travis_clean['log_income'] = np.log(travis_clean['median_income'])

# generating spatial weights 
w = weights.distance.KNN.from_dataframe(travis_clean, k=8)
# row-standardization
w.transform = "R"

# Running OLS Model 
ind_variables = ['distance_to_nearest_uf', 
                 'distance_to_nearest_grocery',
                 'distance_to_nearest_fm',
                 'log_income',
                 'pct_no_vehicle',
                 'median_age',
                 'walking_ind'
                ]

ols = spreg.OLS(
    # Dependent variable
    travis_clean[["E_DIABETES"]].values,
    # Independent variables
    travis_clean[ind_variables].values,
    # Dependent variable name
    name_y="pct_diabetes",
    # Independent variable name
    name_x=ind_variables,
)
print(ols.summary)

# generating moran's residuals 
moran_residuals = spreg.MoranRes(ols, w, z=True)
print(f"Observed Moran's I:  {moran_residuals.I:.4f}")
print(f"Expected Moran's I:  {moran_residuals.eI:.4f}")
print(f"Standardized Z-score: {moran_residuals.zI:.4f}")
print(f"Analytical P-value:  {moran_residuals.p_norm:.4f}")

# generating LM tests 
lm_results = spreg.LMtests(ols, w)
print("LM Error p-value:", lm_results.lme[1])
print("LM Lag p-value:", lm_results.lml[1])
```

```python


# ---------------------------------------------------
# 1. Construct 'y' (Dependent Variable)
# ---------------------------------------------------
# spreg expects y to be an N x 1 column vector. 
# Extracting a single pandas column returns a 1D array, so we must reshape it.

y_col = 'E_DIABETES'
y = travis_clean[y_col].values.reshape(-1, 1)

# ---------------------------------------------------
# 2. Construct 'X' (Independent Variables/Covariates)
# ---------------------------------------------------
# spreg expects X to be an N x K matrix. 
# Extracting multiple columns naturally creates a 2D array.

x_cols = ['distance_to_nearest_uf', 'distance_to_nearest_grocery', 'distance_to_nearest_fm','log_income','pct_no_vehicle','median_age','walking_ind']
X = travis_clean[x_cols].values

# (Optional) Check your shapes to ensure N is consistent
print(f"Shape of y: {y.shape}")  # Should be (N, 1)
print(f"Shape of X: {X.shape}")  # Should be (N, K)

# Assuming 'y' is an N x 1 array, 'X' is an N x K array (without a constant),
# and 'w' is a libpysal spatial weights object.

# 1. Unrestricted Model: Spatial Durbin Model (SDM)
sdm = ML_Lag(y, X, w, slx_lags=1)
ll_sdm = sdm.logll

# 2. Restricted Model A: Spatial Lag Model (SAR)
sar = ML_Lag(y, X, w)
ll_sar = sar.logll

# 3. Restricted Model B: Spatial Error Model (SEM)
sem = ML_Error(y, X, w)
ll_sem = sem.logll

# 4. Likelihood Ratio (LR) Tests
# The degrees of freedom equal the number of exogenous variables K in your X matrix
df = X.shape[1] 

# LR Test for H0: theta = 0 (SDM vs SAR)
lr_stat_sar = 2 * (ll_sdm - ll_sar)
p_val_sar = chi2.sf(lr_stat_sar, df)

# LR Test for H0: theta + rho*beta = 0 (SDM vs SEM)
lr_stat_sem = 2 * (ll_sdm - ll_sem)
p_val_sem = chi2.sf(lr_stat_sem, df)

print(f"LR Test (SDM vs SAR): Stat = {lr_stat_sar:.4f}, p-value = {p_val_sar:.4e}")
print(f"LR Test (SDM vs SEM): Stat = {lr_stat_sem:.4f}, p-value = {p_val_sem:.4e}")
```

Results of LR tests suggest an SEM.

Results suggest an SDM. Given that SDM will collapse into either the SAR or SEM, for the sake of model consistency, I will run an SDM for all counties.
